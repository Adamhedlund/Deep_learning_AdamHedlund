# L02 – Lösningar

```python
import random
import numpy as np
import torch

def set_seed(seed: int = 42) -> None:
    random.seed(seed); np.random.seed(seed)
    torch.manual_seed(seed); torch.cuda.manual_seed_all(seed)
```

### Övning 1

```python
batch = torch.rand(16, 3, 32, 32)
flat_batch = batch.reshape(16, -1)   # eller batch.flatten(start_dim=1)
assert flat_batch.shape == (16, 3072)
print("Rätt!")
```

### Övning 2

```python
set_seed(1)
M = torch.rand(100, 4) * torch.tensor([1.0, 10.0, 100.0, 1000.0])
M_std = (M - M.mean(dim=0)) / M.std(dim=0)
assert torch.allclose(M_std.mean(dim=0), torch.zeros(4), atol=1e-5)
assert torch.allclose(M_std.std(dim=0), torch.ones(4), atol=1e-5)
print("Rätt!")
```

### Övning 3

```python
w = torch.tensor(2.0, requires_grad=True)
g = w**3 - 2*w
g.backward()
grad_value = w.grad.item()
assert abs(grad_value - 10.0) < 1e-6
print("Rätt!", grad_value)
```

### Övning 4

```python
a = torch.tensor(0.0, requires_grad=True)
b = torch.tensor(0.0, requires_grad=True)
lr = 0.1
for step in range(100):
    loss = (a - 5)**2 + (b + 2)**2
    loss.backward()
    with torch.no_grad():
        a -= lr * a.grad
        b -= lr * b.grad
    a.grad.zero_()
    b.grad.zero_()

assert abs(a.item() - 5) < 1e-2 and abs(b.item() + 2) < 1e-2
print("Rätt!", a.item(), b.item())
```
